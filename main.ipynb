```python
import pandas as pd
df = pd.read_csv('population.csv')
df.head()
```

```python
highest = df['2020'].nlargest(20).tolist()
filtered_highest = df[df['2020'].isin(highest)]
filtered_highest
```

```python
import seaborn as sns
sns.barplot(x='Country Name', y='2020', data=df.tail())
```

```python
world=df.loc[df['Country Name']=='World']
world.drop(['Country Name','Country Code',
'Indicator Name','Indicator Code'],axis=1,inplace=True)
world = world.T
world.head()
```

```python
world.dropna(inplace=True)
world=world.reset_index().rename(columns={259:'population',
'index':'year'})
world.tail()
```

```python
sns.lineplot(x='year', y='population', data=world.tail(10))
sns.scatterplot(x='year', y='population', data=world.tail(10))
```

```python
world['population'] = pd.to_numeric(world['population'])
world['year'] = pd.to_numeric(world['year'])
x = world['year']
y = world['population']
```

```python
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, 
test_size=0.2)
X_train = X_train.values.reshape(-1,1)
X_test = X_test.values.reshape(-1,1)
y_train = y_train.values.reshape(-1,1)
y_test = y_test.values.reshape(-1,1)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model = model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_pred
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("True vs Predicted ")
sns.lineplot(x=X_test.reshape(-1), y=y_pred.reshape(-1))
sns.scatterplot(x=X_test.reshape(-1), y=y_pred.reshape(-1), marker="x", color="green", s=200) #Untuk nilai prediksi disimbolkan dengan Cross(x) hijau
sns.scatterplot(x=X_test.reshape(-1), y=y_test.reshape(-1), color="red") # Untuk nilai sesungguhnya disimbolkan dengan scatterplot merah
```

```python
test_data = model.predict([[2030]]) #Input yang dimasukannya harus dalam bentuk 2D array
test_data
```

```python
year_input = int(input("Year : "))
predict_input = model.predict([[year_input]])

print(f"World Population Prediction in {year_input} is {int(predict_input[0])}")
```
